```python
print("Lead_Tag :",df.lead_tag.unique())
print("Lead_Tag :",df1.lead_tag.unique())
print("Lead_Stage :",df1.lead_stage.unique())
print("Lead_Status :",df1.lead_status.unique())
```

Since the missing values in Lead Tag is not only associated with first call empty field we can fill it with those empty cells having corresponding 'Converted' with 'already Purchased.
Lead stage : Follow up required -> Lead Tag : Warm
Lead stage : prospect-> lead tag : warm
Lead stage : Lost -> lead tag : lost to competition

Lead status : Pending  with executive,open -> lead tag : call not connected
Lead status : RESCHEDULED BY EXECUTIVE -> lead tag : warm


```python
df.loc[(df['lead_stage']=='CONVERTED')&(df['lead_tag'].isna()),'lead_tag']='already-purchased'
```

```python
df.loc[((df['lead_stage']=='Follow Up Required')|(df['lead_stage']=='Prospect'))&(df['lead_tag'].isna()),'lead_tag']='warm'
```

```python
df.loc[(df['lead_stage']=='LOST')&(df['lead_tag'].isna()),'lead_tag']='lost-to-competition'
```

```python
print(df.isna().sum())
```

With mapping lead stage we were able to fill the whole empty cells under lead tag and now the only remaining empty cells is under first call attempt,last call dtae and alst call time. These fields may be missed and is only a time stamps. Hence it may not effect the ML model  for predicting the lead connvertibility and can drop those columns as well.

```python
df=df.drop(["first_call_attempt","last_call_date","last_call_time"],axis=1)
print(df)
print(df.isna().sum())
print(df.dtypes)
print(df.shape)
```

```python
df=df.drop_duplicates()
print(df)
```

No duplicates present. The fields Name,Contact no, Contact email, Creation date may not be useful for creating the models and we can drop those as well.

```python
df=df.drop(["Name","contact_number","Contact_email","creation_date"],axis=1)
df
```

Among the data the lead stage can be considered as output so that other parameters can be changed like inouts are sources like source type, source name, campaign name with control variables like assigned to and derived inputs like lead tag and lead status. Hence we can encode using get dummy method.

```python
dummy=pd.get_dummies(df[['source_name','source_type','campaign_name','assigned_to','lead_tag','lead_status']],drop_first=True)
```

```python
dfn=pd.concat([df,dummy],axis=1)
dfn
```

```python
dfn=dfn.drop(["lead_status","lead_tag","assigned_to","campaign_name","source_type","source_name"],axis=1)
dfn
```

```python
x=dfn.iloc[:,1:]
y=dfn.iloc[:,0]
y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20)
x_test
```

```python
from sklearn.tree import DecisionTreeClassifier
dt=DecisionTreeClassifier()
dt.fit(x_train,y_train)
y_pred=dt.predict(x_test)
```

```python
from sklearn.metrics import classification_report,confusion_matrix,accuracy_score
print(confusion_matrix(y_pred,y_test))
print(classification_report(y_pred,y_test))
```
